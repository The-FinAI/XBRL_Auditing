```python
def find_chunk_by_concept(jsonl_file, query_concept_id):
    """
    给定一个 us-gaap concept_id，在 jsonl 文件中找到完全匹配的条目，
    并返回对应的 chunk_text
    :param jsonl_file: jsonl 文件路径
    :param query_concept_id: 例如 "us-gaap:CashAndDueFromBanks"
    :return: list of chunk_text
    """
    results = []
    with open(jsonl_file, "r", encoding="utf-8") as f:
        for line in f:
            try:
                obj = json.loads(line)
                # print(obj)
                if obj.get("concept_id") == query_concept_id:  # 完全匹配
                    # print(1)
                    results.append(obj.get("chunk_text"))
            except json.JSONDecodeError:
                continue
    return results
```

```python
def extract_usgaap_segmentation(df, task_name="RE"):
    new_df = df.copy(deep=True)
    for idx in tqdm(range(len(new_df))):
        report_year = str(new_df.at[idx, "report.filing-year"])
        is_parse = new_df.at[idx, "parse_successful"]
        prefix_path = index_prefix[report_year]
        concept_file = os.path.join(prefix_path,"chunks_core.jsonl")
        relation_file = os.path.join(prefix_path, "chunks_relations.jsonl")
        query = []
        if task_name == "MR":
            query_str = new_df.at[idx, "input"]
            query_json = eval(query_str)
            query.append(query_json.get("tag"))
        elif task_name == "RE":
            query_str = new_df.at[idx, "true_answer"]
            query_json = eval(query_str)
            for a in query_json:
                query.extend([a[0], a[-1]])
        elif task_name == "SM":
            query_str = new_df.at[idx, "true_answer"]
            query_json = eval(query_str)
            for a in query_json:
                if "=" in a:
                    query.extend(a.split("="))
                else:
                    query.append(a)
        else:
            raise ValueError("The task name must be MR, RE, or SM.")
        # print(query)
        segments = []
        for q in query:
            # print(q)
            res = find_chunk_by_concept(concept_file,q)
            segments.extend(res)
        for q in query:
            res = find_chunk_by_concept(relation_file,q)
            segments.extend(res)

        new_df.at[idx, "us_gaap"] = "\n\n\n".join(segments)

    return new_df
            
        
```

```python
new_df = extract_usgaap_segmentation(initial_task_df, task_name="MR")
```
