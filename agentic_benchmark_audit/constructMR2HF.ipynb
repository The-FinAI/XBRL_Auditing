```python
system_prompt_MR = """You are an auditor for XBRL filings. 
Given the question and the provided filing 
(schema, presentation, calculation, definition, label, instance, and US GAAP taxonomy), 
identify the reported value of a financial element and calculate the actual value that should be reported based on calculation relationships.

Answer strictly in the following JSON format:
{
  "extracted_value": "<numeric value reported in the instance document, or 0 if not found, keep the same format as in the XBRL filing>",
  "calculated_value": "<numeric value computed from calculation relationships, or 0 if not computable, use the same number formatting style as extracted_value>"
}

No explanations. No extra text. Only the JSON object.
"""
```

```python
def get_user_prompt(question1, question2, schema, presentation, calculation, definition, label, instance, taxonomy):
    prompt = f"""
##Schema document
{schema}

##Presentation linkbase document
{presentation}

##Calculation linkbase document
{calculation}

##Definition linkbase document
{definition}

##Label linkbase document
{label}

##Instance document
{instance}

##US GAAP Taxonomy
{taxonomy}

Question1: {question1}
Question2: {question2}
Answer:
"""

    return prompt
```

```python
query = []
answer = []
dqc_id = []
for row in tqdm(df_0015.itertuples()):
    schema = row.schema 
    if pd.isna(schema):
        schema = "None"

    presentation = row.presentation
    if pd.isna(presentation):
        presentation = "None"

    calculation = row.calculation
    if pd.isna(calculation):
        calculation = "None"

    definition = row.definition
    if pd.isna(definition):
        definition = "None"

    label = row.label
    if pd.isna(label):
        label = "None"

    instance = row.instance
    if pd.isna(instance):
        instance = "None"

    taxonomy = row.us_gaap
    if pd.isna(taxonomy):
        taxonomy = "None"

    input = eval(row.input)
    
    question1 = f"""What's the reported value of {input.get('tag')} in the Instance document for the period {input.get('period')}?"""
    question2 = f"""What's the actual value of {input.get('tag')} in the Instance document for the period {input.get('period')}, calculated based on the calculation relationship?"""
    user_input = get_user_prompt(question1,question2,schema,presentation,calculation,definition,label,instance,taxonomy)
    
    q = f"""{system_prompt_MR}
{user_input}
"""
    a = eval(row.true_answer)
    a = {"extracted_value": a.get("extracted_value"), "calculated_value": a.get("calculated_value")}
    id = "DQC_US_0015"
    
    query.append(q)
    answer.append(a)
    dqc_id.append(id)
```
